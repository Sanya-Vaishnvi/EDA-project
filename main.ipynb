# Project part 1: Data Collection
We  noticed that the url is of the form:\
`https://portal.amfiindia.com/DownloadNAVHistoryReport_Po.aspx?mf=&frmdt=13-Jan-2015&todt=01-Mar-2015&tp=1`\

from visual inspection, we can see that the url calls a file from a `<from_data>` to a `<to_date>`

```python
%pip install requests pandas seaborn
import requests
import os
import pandas as pd
import seaborn as sb
```

```python
number_years_required = 10
start_year = 2024-number_years_required

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

data_folder = "downloads"

if not os.path.exists(data_folder):
    os.makedirs(data_folder)

for i in range(number_years_required):
    for j in range(0, 12, 2):
        from_month = months[j]
        to_month = months[(j + 2)%12]

        from_date = f"02-{from_month}-{start_year + i}"
        to_date = f"01-{to_month}-{start_year + i}"

        if to_month == "Jan":
            to_month = "Dec"
            to_date = f"31-{to_month}-{start_year + i}"
        
        if from_month == "Jan":
            from_date = f"01-{from_month}-{start_year + i}"


        print(f"From: {from_date}, To: {to_date}")

        if os.path.exists(f"{data_folder}/{start_year+i}_{j}.csv"):
            print("File already exists, skipping it")
            continue

        

        url = f"https://portal.amfiindia.com/DownloadNAVHistoryReport_Po.aspx?mf=&frmdt={from_date}&todt={to_date}&tp=1"
        response = requests.get(url, timeout=15)

        # raw HTML/text
        html = response.text

        with open(f"{data_folder}/{start_year+i}_{j}.csv", "w", encoding="utf-8") as f:
            f.write(html)
            print(f"Saved {from_date}_to_{to_date}.csv")
```

# Part 2: Filtering Data
We need the ELSS data, not the other categories mentioned in the AMFI india website. 
We have already used the filter on the site to get the "open eneded" type.

```python
with open("final_data.csv", "w") as data:    
    for file_name in os.listdir(data_folder):
        print(file_name)
        with open(f"downloads/{file_name}", "r", encoding="UTF8") as file:
            log = 0
            for line in file:
                if "Open Ended Schemes ( ELSS )" in line:
                    print("Started Logging")
                    log = 1
                elif log and "Open Ended Schemes ( " in line:
                    break
                elif log:
                    if len(line) > 1:
                        split = line.split(";")
                        if split[0].isdecimal():
                            data.write(",".join(split))
```
